```python
total_sales = df['Sales'].sum()

print("Total Sales:", total_sales)
```

```python
total_profit = df['Profit'].sum()

print("Total Profit:", total_profit)
```

```python
total_quantity = df['Quantity'].sum()

print("Total Quantity:", total_quantity)
```

```python
category_sales = df.groupby('Category')['Sales'].sum()

print(category_sales)
```

```python
category_profit = df.groupby('Category')['Profit'].sum()

print(category_profit)
```

```python
category_sales.sort_values(ascending=False)
```

```python
plt.figure(figsize=(8,5))

category_sales.sort_values().plot(
    kind='bar',
    title='Category-wise Sales Analysis'
)

plt.xlabel('Category')
plt.ylabel('Total Sales')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8,5))

category_profit.sort_values().plot(
    kind='bar',
    title='Category-wise Profit Analysis'
)

plt.xlabel('Category')
plt.ylabel('Total Profit')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
region_sales = df.groupby('Region')['Sales'].sum()

print(region_sales)
```

```python
region_profit = df.groupby('Region')['Profit'].sum()

print(region_profit)
```

```python
plt.figure(figsize=(8,5))

region_sales.sort_values().plot(
    kind='bar',
    title='Region-wise Sales Analysis'
)

plt.xlabel('Region')
plt.ylabel('Total Sales')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8,5))

region_profit.sort_values().plot(
    kind='bar',
    title='Region-wise Profit Analysis'
)

plt.xlabel('Region')
plt.ylabel('Total Profit')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
product_sales = df.groupby('Sub-Category')['Sales'].sum()

print(product_sales.sort_values(ascending=False))
```

```python
top_10_products = product_sales.sort_values(
    ascending=False
).head(10)

print(top_10_products)
```

```python
plt.figure(figsize=(10,6))

top_10_products.sort_values().plot(
    kind='barh',
    title='Top 10 Products by Sales'
)

plt.xlabel('Total Sales')
plt.ylabel('Sub-Category')
plt.tight_layout()
plt.show()
```

```python

product_profit = df.groupby('Sub-Category')['Profit'].sum()

top_10_profit = product_profit.sort_values(
    ascending=False
).head(10)

print(top_10_profit)
```

```python
plt.figure(figsize=(10,6))

top_10_profit.sort_values().plot(
    kind='barh',
    title='Top 10 Products by Profit'
)

plt.xlabel('Total Profit')
plt.ylabel('Sub-Category')
plt.tight_layout()
plt.show()
```

```python
loss_products = df.groupby('Sub-Category')['Profit'].sum()

loss_products = loss_products[loss_products < 0]

print(loss_products)
```

```python
plt.figure(figsize=(9,5))

loss_products.sort_values().plot(
    kind='bar',
    title='Loss Making Sub-Categories'
)

plt.xlabel('Sub-Category')
plt.ylabel('Total Loss')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
